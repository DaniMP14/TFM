# Exploracion de una noche de observacion

Esta libreta sirve para inspeccionar una noche completa, contar FITS por tipo y revisar cabeceras antes de lanzar la reduccion.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()
SRC_DIR = PROJECT_ROOT / 'src'
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from reduction.io import group_by_frame_type, index_frames
```

```python
raw_dir = PROJECT_ROOT / 'data' / 'raw'
records = index_frames(raw_dir)
grouped = group_by_frame_type(records)
summary = {key: len(value) for key, value in grouped.items()}
summary
```

```python
for frame_type, frame_records in grouped.items():
    print(f'\n[{frame_type}] {len(frame_records)} archivos')
    for record in frame_records[:5]:  # Mostrar solo los primeros 5 registros de cada tipo
        print(record)
```
